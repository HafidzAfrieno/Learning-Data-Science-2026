```python
import torch
import numpy as np
```

## 1. Menginisialisasi Tensor

```python
data = [[1, 2],[3, 4],[2,8]]
np_array = np.array(data)

x_np = torch.from_numpy(np_array)
x_np
```

```python
x_data = torch.tensor(data)
x_data
```

```python
#pakai variabel x_np
x_ones_ts = torch.ones_like(x_np)
x_rand_ts = torch.rand_like(x_np, dtype=torch.float)

print(f"Ones Tensor: \n {x_ones_ts} \n")
print(f"Random Tensor: \n {x_rand_ts} \n")
```

```python
shape = (3,5)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```
