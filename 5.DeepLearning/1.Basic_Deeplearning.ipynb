## 1. Basic Math DeepLearning

```python
import numpy as np
import pandas as pd
```

```python
class ModelRegresi:
    def __init__(self, learning_rate=0.1,weight=0.0,bias=0.0):
        """Inisialisasi parameter model regresi linear"""
        self.weight = weight
        self.bias = bias
        self.learning_rate = learning_rate

    def function_linear(self, X_batch):
        """Menghitung Prediksi: (w * x) + b"""
        return (self.weight * X_batch) + self.bias

    def function_error(self, y_predict, y_batch):
        """Menghitung Error: Prediksi - Target"""
        return y_predict - y_batch

    def function_gradient(self, X_batch, error):
        """Menghitung rata-rata gradien w dan b untuk minibatch"""
        batch_size = len(X_batch)
        gradient_weight = error * X_batch
        gradient_bias = error
      
        mean_gradient_weight = np.sum(gradient_weight) / batch_size
        mean_gradient_bias = np.sum(gradient_bias) / batch_size
        return mean_gradient_weight, mean_gradient_bias

    def function_update_params(self, X_batch, y_batch):
        """Proses pembaharuan bobot dan bias untuk satu minibatch"""
        y_predict = self.function_linear(X_batch)
        error = self.function_error(y_predict, y_batch)
        
        gw, gb = self.function_gradient(X_batch, error)
        self.weight -= self.learning_rate * gw
        self.bias -= self.learning_rate * gb

    def fit_minibatch(self, X_feature, y_target, batch_size=2, epochs=1):
        """Proses training dengan Minibatch SGD"""
        X = np.array(X_feature)
        y = np.array(y_target)
        n_samples = len(X)

        for epoch in range(epochs):
            # Iterasi per minibatch
            for i in range(0, n_samples, batch_size):
                X_batch = X[i : i + batch_size]
                y_batch = y[i : i + batch_size]
                self.function_update_params(X_batch, y_batch)

    def predict(self, X_feature):
        """Menghasilkan array output prediksi"""
        return self.function_linear(np.array(X_feature))

    def function_metrik_MSE(self, X_feature, y_target):
        """Evaluasi model menggunakan Mean Squared Error (MSE)"""
        y_predict = self.function_linear(np.array(X_feature))
        error = y_predict - np.array(y_target)
        return np.mean(error ** 2)
```

```python
def tampilkan_evaluasi(judul, model, X, y):
    """Fungsi pembantu untuk menampilkan status model dan tabel prediksi."""
    prediksi = model.predict(X)
    mse = model.function_metrik_MSE(X, y)

    print("=" * 45)
    print(f"=== {judul.upper()} ===")
    print("=" * 45)
    print(f"Bobot (w) : {model.weight:.4f}")
    print(f"Bias (b)  : {model.bias:.4f}")
    print(f"MSE       : {mse:.4f}")
    print("-" * 45)
    print(f"{'Input (x)':<12} | {'Target (y)':<12} | {'Prediksi':<12}")
    print("-" * 45)

    for x_val, y_val, y_pred in zip(X, y, prediksi):
        print(f"{x_val:<12} | {y_val:<12} | {y_pred:<12.4f}")
    print("\n")
```

```python
X = [1, 2, 3, 4]
y = [3, 5, 7, 9]
model = ModelRegresi(learning_rate=0.1,weight=0.0,bias=0.0)

tampilkan_evaluasi("Sebelum Training", model, X, y) # 1. Evaluasi Sebelum Training

model.fit_minibatch(X, y, batch_size=2, epochs=1) # 2. Proses Training (1 Epoch, Batch Size = 2)

tampilkan_evaluasi("Setelah 1 Epoch Training", model, X, y) # 3. Evaluasi Setelah Training
```
